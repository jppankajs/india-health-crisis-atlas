```python
import pandas as pd
import numpy as np 
import warnings
warnings.filterwarnings('ignore')

print("Libraries loaded successfully")
print(f"pandas version: {pd.__version__}")
print(f"numpy version: {np.__version__}")
```

```python
#load the dataset
#'..' means go one folder up from notebooks/ to reach the project roots
#then go into data/raw/ to find the file

df = pd.read_excel(
    '../data/raw/NFHS_5_India_Districts_Factsheet_Data.xls',
    engine= 'xlrd',
    header=0
)

print("Dataset loaded successfully!")
print(f"shape: {df.shape}")
```

```python
#describe() but readable -only show key colums
#round to 2 decimal places for clean reading

df.describe().round(2).T
```

```python
#check how many missing values in each column

missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
     'missing count':missing,
     'missing %': missing_pct.round(2)
})

#show only columns that have missing values
print(missing_report[missing_report['missing count']>0])
print(f"\n total coulmns with missing data: {(missing >0).sum()}")
```
